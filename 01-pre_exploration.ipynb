# PRE EXPLORATION DATASET

In this jupyter notebook i'm going to take a first observation of the dataset, set them a little and understand what type of informations they contains. <br>
The questions we will analyze in this analysis are: <br>
- Trade openness and employment: Do EU countries with greater trade openness (exports as a share of GDP) show different unemployment levels?
- Trade composition and employment: Is the weight of manufactured goods in a country's trade associated with different unemployment levels?
- Reaction to economic shocks: How did trade and unemployment react to the 2009 and 2020 crises, and which countries recovered fastest?

```python
GDP_FOR_NATION_path= "data/raw_data/gdp_for_nation.csv"
DISOCCUPATION_FOR_NATION_path= "data/raw_data/disoccupation_for_nation.csv"
IN_EX_EU_TRADE_path= "data/raw_data/in_ex_eu_trade.tsv"
IN_EX_EU_TRADE_MANUFACTURED_path= "data/raw_data/in_ex_eu_trade_manufactured.tsv"
```

```python
GDP_FOR_NATION= pd.read_csv(GDP_FOR_NATION_path)
DISOCCUPATION_FOR_NATION= pd.read_csv(DISOCCUPATION_FOR_NATION_path)
IN_EX_EU_TRADE=  pd.read_csv(IN_EX_EU_TRADE_path, sep="\t")
IN_EX_EU_TRADE_MANUFACTURED= pd.read_csv(IN_EX_EU_TRADE_MANUFACTURED_path, sep= "\t")
```

```python
with open(IN_EX_EU_TRADE_path) as f:
    for _ in range(3):
        print(repr(f.readline()))
```

```python
IN_EX_EU_TRADE.head(5)
```

```python

IN_EX_EU_TRADE_indicator= IN_EX_EU_TRADE[IN_EX_EU_TRADE.columns[0]].str.split(",", expand=True)
```

```python
IN_EX_EU_TRADE_indicator.head(2)
```

```python
IN_EX_EU_TRADE_indicator.rename(columns={
    0: "freq",
    1: "indic_et",
    2: "sitc06",
    3: "partner",
    4: "geo"
}, inplace= True)
```

```python
IN_EX_EU_TRADE_nocol0= IN_EX_EU_TRADE.drop(["freq,indic_et,sitc06,partner,geo\TIME_PERIOD"], axis=1)
```

```python
IN_EX_EU_TRADE_nocol0.head(1)
```

```python
IN_EX_EU_TRADE=pd.concat([IN_EX_EU_TRADE_indicator,IN_EX_EU_TRADE_nocol0], join= "outer", axis=1)
```

```python
IN_EX_EU_TRADE.head(2)
```

```python
IN_EX_EU_TRADE_melted=IN_EX_EU_TRADE.melt(
    id_vars= ["freq", "indic_et", "sitc06", "partner", "geo"],
    var_name= "time_period",
    value_name= "OBS_VALUE"
)
```

```python
IN_EX_EU_TRADE_melted.head(5)
```

```python
IN_EX_EU_TRADE_melted[
    (IN_EX_EU_TRADE_melted["sitc06"] == "SITC0_1") & (IN_EX_EU_TRADE_melted["geo"] == "AT")
]
```

```python
IN_EX_EU_TRADE_melted["indic_et"].value_counts()
```

This DS "IN_EX_EU_TRADE_melted contains: <br>
- MIO_EXP_VAL: exports in millions of euros
- MIO_IMP_VAL: imports in millions of euros
- MIO_BAL_VAL:difference between exports and imports

- PC_EXP_PROD: how much a product's category weighs on the total exports percentage
- PC_IMP_PROD:how much a product's category weighs on the total  imports percentage

- CONT_EXP_EU: share of exports to the EU
- CONT_IMP_EU: share of imports from EU

- PC_TOT_EXP: share of total exports in %
- PC_TOT_IMP: share of total imports in %

```python
IN_EX_EU_TRADE_melted.to_csv("data/semi-processed_data/IN_EX_EU_TRADE_melted.csv", index= False)
```
