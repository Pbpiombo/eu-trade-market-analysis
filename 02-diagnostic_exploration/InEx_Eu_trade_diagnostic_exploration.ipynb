```python
#filter the indic_et
list_indic= ["MIO_BAL_VAL","MIO_EXP_VAL", "MIO_IMP_VAL"]
in_ex_EU_world=in_ex_EU_world[ in_ex_EU_world["indic_et"].isin(list_indic)]
```

I will explain more clearly the "indic_et" values and the "sitc06" values, before proceding with the creation of the diffent df. <br>
 I will rename the firsts and create a a product-clarification column for the seconds.product_code
 I also rename the columns to be consistent with the previous dataframes. <br>
 i will not create a "country" column, because i will use these dataframe principally for join.

```python
in_ex_EU_world["indic_et"]= in_ex_EU_world["indic_et"].replace({"MIO_BAL_VAL": "balance", "MIO_EXP_VAL": "export", "MIO_IMP_VAL": "import"})
```

```python
in_ex_EU_world["indic_et"].value_counts()
```

```python
product_labels = {
    "SITC0_1": "food_drinks_tobacco",
    "SITC2_4": "raw_materials",
    "SITC3": "mineral_fuels",
    "SITC5": "chemicals",
    "SITC6_8": "other_manufactured",
    "SITC7": "machinery_transport",
    "SITC9": "unclassified",
    "TOTAL": "total"
}
in_ex_EU_world["product_description"] = in_ex_EU_world["sitc06"].map(product_labels)
```

```python
in_ex_EU_world["product_description"].value_counts()
```

```python
in_ex_EU_world= in_ex_EU_world.rename(columns={"sitc06":"product_code", "geo": "country_code", "time_period": "year" , "OBS_VALUE": "value_meur"})
```

```python
#change the order of the columns:as
col_order = ["country_code","product_code", "product_description", "year", "indic_et", "value_meur"]
in_ex_EU_world=in_ex_EU_world[col_order]
```

Now i create two diffents dataset: one for product categories and one for total.

```python
product_list= ["SITC0_1","SITC2_4","SITC3","SITC5","SITC6_8","SITC7", "SITC9"]
in_ex_EU_prcategories= in_ex_EU_world[in_ex_EU_world["product_code"].isin(product_list)].copy()
in_ex_EU_total= in_ex_EU_world[in_ex_EU_world["product_code"] =="TOTAL"].copy()
```
