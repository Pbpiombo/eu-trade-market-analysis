Before delving into the "OBS_FLAG" values, I isolate the "Percentage of population in the labour force," which is the standard metric for defining unemployment and the only one that allows for cross-country comparison. <br>
For the sake of clarity, I am specifying the meaning of the "OBS_FLAG":
- b: It changes the standard by which a variable is defined and observed over time, as a result of a single change or a combination of multiple changes at a given moment.
- d: The definition used deviates from the standard, with details in the metadata. 
- db: A break in the series and a differing definition. The three BD values ​​possess both characteristics.


```python
DISOCCUPATION_clean= DISOCCUPATION_FOR_NATION[DISOCCUPATION_FOR_NATION["unit"].isin(["Percentage of population in the labour force"])]
```

```python
DISOCCUPATION_clean.head(5)
```

Now i will isolate in a new df the EU27-2020 value and create a new df with the EU 27 countries only.

And concentrate the database only on the 27 EU countries

```python
list_27_names = [
    "Belgium", "Bulgaria", "Czechia", "Denmark", "Germany", "Estonia",
    "Ireland", "Greece", "Spain", "France", "Croatia", "Italy", "Cyprus",
    "Latvia", "Lithuania", "Luxembourg", "Hungary", "Malta", "Netherlands",
    "Austria", "Poland", "Portugal", "Romania", "Slovenia", "Slovakia",
    "Finland", "Sweden"
]
disoccupation_eu_countries = DISOCCUPATION_clean[DISOCCUPATION_clean["geo"].isin(list_27_names)].copy()
```

Now I'm creating a column with country codes, which will be useful for future joins and cross-analysis.


```python
country_to_code = {
    "Austria": "AT",
    "Belgium": "BE",
    "Bulgaria": "BG",
    "Croatia": "HR",
    "Cyprus": "CY",
    "Czechia": "CZ",
    "Denmark": "DK",
    "Estonia": "EE",
    "Finland": "FI",
    "France": "FR",
    "Germany": "DE",
    "Greece": "EL",
    "Hungary": "HU",
    "Ireland": "IE",
    "Italy": "IT",
    "Latvia": "LV",
    "Lithuania": "LT",
    "Luxembourg": "LU",
    "Malta": "MT",
    "Netherlands": "NL",
    "Poland": "PL",
    "Portugal": "PT",
    "Romania": "RO",
    "Slovakia": "SK",
    "Slovenia": "SI",
    "Spain": "ES",
    "Sweden": "SE"
}
disoccupation_eu_countries["country_code"]=disoccupation_eu_countries["geo"].map(country_to_code)
```


Several flagged records remain; let's run a check to see for which countries and years the change occurred.

To avoid misinterpreting the data in subsequent analyses, i decided to save a mini-dataframe containing information on the countries and the year in which a change was applied to the unemployment analysis metrics.

```python
unemployment_data_breaks=disoccupation_eu_countries[disoccupation_eu_countries["OBS_FLAG"].notna()].copy()
```

```python
unemployment_data_breaks
```

From this dataframe, I will keep only country, country_code, OBS_FLAG and year. <br>
It will serve only to remind me which countries changed their measurement metrics, in which year and in what way.

```python
unemployment_data_breaks=unemployment_data_breaks.drop(["OBS_VALUE"], axis=1)
unemployment_data_breaks
```

```python
flag_labels = {
    "b": "break_in_series",
    "d": "definition_differs",
    "bd": "break_and_definition"
}
unemployment_data_breaks["OBS_FLAG"]=unemployment_data_breaks["OBS_FLAG"].replace(flag_labels)
```

```python
unemployment_data_breaks
```

Now I can remove the "OBS_FLAG" col from the disoccupation_eu_countries DataFrame.


```python
disoccupation_eu_countries=disoccupation_eu_countries.drop(["OBS_FLAG"], axis=1)
```

```python
disoccupation_eu_countries
```

 Now i rename the cols of the df, to be in line with GDP_EU_countries df done previously. <br>
 after that i reorder the col to make reading easier and save it in clean_data togheter with unemployment_data_breaks df.

```python
disoccupation_eu_countries=disoccupation_eu_countries.rename(columns={
     "OBS_VALUE": "unemployment_rate",
    "TIME_PERIOD": "year",
})
```

```python
col_order= ["country_code","geo","year","unemployment_rate"]
disoccupation_eu_countries=disoccupation_eu_countries[col_order]
disoccupation_eu_countries
```

```python
disoccupation_eu_countries=disoccupation_eu_countries.rename(columns={"geo": "country"})
unemployment_data_breaks=unemployment_data_breaks.rename(columns={"geo": "country"})
```
