Now i will isolate in a new df the EU27-2020 value and create a new df with the EU 27 countries only.

```python
gdp_eu27 = GDP_clean[GDP_clean["geo"] == "EU27_2020"]
```

```python
gdp_eu27
```

```python
list_27=["AT","BE","BG","CY","CZ","DE","DK","EE","EL","ES","FI","FR","HR","HU","IE","IT","LT","LU","LV","MT","NL","PL","PT","RO","SE","SI","SK"]
gdp_eu_countries = GDP_clean[GDP_clean["geo"].isin(list_27)].copy()
```

```python
gdp_eu_countries
```

```python
gdp_eu_countries["geo"].value_counts()
```

Finally:

- Rename the 'unit' elements to clarify that one value accounts for inflation while the other does not, thereby eliminating the 'unit of measure' column, which only creates redundancy;
- Rename columns containing spaces that complicate SQL analysis;
- Assign the unit of measure (e.g., million_euro) directly to OBS_VALUE so the information is not lost.

```python
#Rename "unit" element.
gdp_eu_countries["unit"]=gdp_eu_countries["unit"].replace({"CP_MEUR": "nominal", "CLV10_MEUR": "real"})
```

```python
gdp_eu_countries
```

```python
# Rename columns and drop "Unit of measure"
gdp_eu_countries=gdp_eu_countries.rename(columns={
    "OBS_VALUE": "gdp_meur",
    "TIME_PERIOD": "year",
    "geo": "country_code",
    "Geopolitical entity (reporting)":"country"
}).drop("Unit of measure", axis=1)
```
