## Dimension exploration.
In this section i analyze the cols that i can join in the next queries, the date boundaries that i have and the dimensions of all df.

```python
import sqlite3
from pathlib import Path
import pandas as pd

conn = sqlite3.connect(Path.cwd().parent/"database/ue_market.db")
pd.read_sql("SELECT name FROM sqlite_master WHERE   type= 'table'", conn)
```

```python
pd.read_sql("""
            SELECT 
            COUNT(DISTINCT country_code) AS n_countries,
            COUNT(DISTINCT year) AS n_years,
            MIN(year) AS first_year,
            MAX(year) AS last_year,
            COUNT(*) AS total_rows
            FROM unemployment_ue""",
            conn)
```

```python
pd.read_sql("""
            SELECT 
            COUNT(DISTINCT country_code) AS n_countries,
            COUNT(DISTINCT year) AS n_years,
            COUNT(DISTINCT unit) AS n_units,
            MIN(year) AS first_year,
            MAX(year) AS last_year,
            COUNT(*) AS total_rows
            FROM gdp_ue""",
            conn)
```

```python
pd.read_sql("""
            SELECT 
            COUNT(DISTINCT country_code) AS n_countries,
            COUNT(DISTINCT year) AS n_years,
            COUNT(DISTINCT indic_et) AS indics,
            MIN(year) AS first_year,
            MAX(year) AS last_year,
            COUNT(*) AS total_rows
            FROM trade_total""",
            conn)
```

```python
pd.read_sql("""
            SELECT 
            COUNT(DISTINCT country_code) AS n_countries,
            COUNT(DISTINCT year) AS n_years,
            COUNT(DISTINCT indic_et) AS indics,
            COUNT(DISTINCT product_code) AS n_products,
            MIN(year) AS first_year,
            MAX(year) AS last_year,
            COUNT(*) AS total_rows
            FROM trade_category""",
            conn)
```

The number of countries is indeed 27; the "gdp_ue" dataframe has twice as many rows as the "unemployment_ue" dataframe. This is because "gdp_ue" contains two units in the col: one for real values ​​and one for nominal values. <br>
The "trade_category" and "trade_total" dataframes contain historical data starting from 2002, unlike "gdp_ue" and "unemployment_ue". <br>
For the analysis of the 2008 crisis, I will use only the latter dataframes, given that the others lack data for those years
